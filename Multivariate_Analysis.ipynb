```python
import pandas as pd
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
titanic = pd.read_csv('/content/Titanic-Dataset.csv')
```

```python
titanic.head()
```

```python
flights = sns.load_dataset('flights')
```

```python
flights.head()
```

```python
iris = sns.load_dataset('iris')
```

```python
iris.head()
```

## 1. Scatterplot



```python
sns.scatterplot(x='total_bill', y='tip', data=tips, hue='sex', style = 'smoker')
```

##2. Bar Plot


```python
sns.barplot(x='Pclass', y='Age', data=titanic, hue='Sex')
```

##3. Boxplot

```python
sns.boxplot(x='Pclass', y='Age', data=titanic, hue='Survived')
```

##4. Displot


```python
sns.displot(x='Age', data=titanic, hue='Survived', kind='kde')
```

## 5. Heatmap


```python
titanic.head()
```

```python
pd.crosstab(titanic['Pclass'], titanic['Survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'], titanic['Survived']))
```

```python
titanic.groupby(['Pclass', 'Survived'])['PassengerId'].count()
```

```python
mean_age = titanic['Age'].mean()
print(f'The mean age of passengers on the Titanic is: {mean_age:.2f}')
```

```python
titanic.groupby('Pclass')['Age'].mean().plot(kind='bar')
```

```python
(titanic.groupby('Pclass')['Survived'].mean() * 100).plot(kind='bar')
```

```python
titanic.groupby('Pclass')['Sex'].mean().plot(kind='bar')
```

```python
sns.countplot(x='Pclass', hue='Sex', data=titanic)
```

## 6. Clustarmap

```python
sns.clustermap(pd.crosstab(titanic['SibSp'],titanic['Survived']))
```

##7. Pairplot

```python
iris.head()
```

```python
sns.pairplot(iris,hue='species')
```

## 8. Lineplot

```python
flights.head()
```

```python
flights.groupby('year')['passengers'].sum().reset_index()
```

```python
sns.lineplot(x='year', y='passengers', data=flights)
```

```python
sns.heatmap(flights.pivot_table(index='month', columns='year', values='passengers'))
```
